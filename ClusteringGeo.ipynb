```python
import pandas as pd

##############################################################################
# 1. LOAD THE RAW DATASETS
##############################################################################

# A) "clustering.csv" - Voter-year level, with TurnoutRate, lat/lon, precinct
df_cluster = pd.read_csv("clustering.csv")
# Columns (example): VoterID, Year, TurnoutRate, Residence_Addresses_Latitude,
#                   Residence_Addresses_Longitude, Precinct

# B) "enrollment_at_precinct_level.csv" - Precinct-year level, with Voter_Enrollment, etc.
df_enroll = pd.read_csv("enrollment_at_precinct_level.csv")
# Columns (example): Year, Precinct, Voter_Enrollment, ElectionType, TotalEligible

# C) "long_each_voter_turnout (1).csv" - Voter-election level, with TurnoutPercentage (as strings w/ '%'), demographics, etc.
df_turnout = pd.read_csv("long_each_voter_turnout (1).csv")
# Columns (example):
#   [Unnamed: 0, VoterID, ElectionDate, ElectionType, EligibilityStatus,
#    Eligible, What_Age_at_Election_age_Eligible, Registration_Eligible,
#    TurnoutPercentage, Year_Enrollment, Year, Age_Group, Voters_Gender]

##############################################################################
# 2. CLEAN UP 'TurnoutPercentage' IN df_turnout (REMOVE '%', CONVERT TO NUMERIC)
##############################################################################
df_turnout['TurnoutPercentage'] = df_turnout['TurnoutPercentage'].str.replace('%', '', regex=False)
df_turnout['TurnoutPercentage'] = pd.to_numeric(df_turnout['TurnoutPercentage'], errors='coerce')

##############################################################################
# 3. AGGREGATE THE VOTER-ELECTION DATA (df_turnout) TO (VoterID, Year)
##############################################################################
# For each (VoterID, Year), compute:
#   - Mean TurnoutPercentage -> AggTurnoutRate
#   - First Age_Group, first Voters_Gender, first Year_Enrollment (or any other fields)
df_turnout_agg = (
    df_turnout
    .groupby(['VoterID', 'Year'], as_index=False)
    .agg({
        'TurnoutPercentage': 'mean',   # or 'max' / 'min' / etc.
        'Age_Group': 'first',
        'Voters_Gender': 'first',
        'Year': 'first'  # If this column exists in your data
    })
)

# Rename the aggregated turnout column for clarity
df_turnout_agg.rename(columns={'TurnoutPercentage': 'AggTurnoutRate'}, inplace=True)

##############################################################################
# 4. MERGE AGGREGATED TURNOUT WITH CLUSTER DATA (df_cluster)
##############################################################################
# df_cluster is also at (VoterID, Year) granularity.
# We'll do a left join so we keep all rows from df_cluster
df_merged = pd.merge(
    df_cluster,         # [VoterID, Year, TurnoutRate, lat/lon, Precinct]
    df_turnout_agg,     # [VoterID, Year, AggTurnoutRate, Age_Group, Voters_Gender, ...]
    on=['VoterID', 'Year'],
    how='left'
)

# Decide how to handle the two turnout columns:
#   - "TurnoutRate" from clustering.csv
#   - "AggTurnoutRate" from your new aggregation
# If you want to replace the old TurnoutRate with the new one when available:
df_merged['TurnoutRate'] = df_merged['AggTurnoutRate'].combine_first(df_merged['TurnoutRate'])
df_merged.drop(columns=['AggTurnoutRate'], inplace=True)

##############################################################################
# 5. MERGE PRECINCT-LEVEL DATA (df_enroll) INTO df_merged
##############################################################################
# df_enroll is at (Precinct, Year, ElectionType) level. If you only need totals per precinct-year,
# aggregate across ElectionTypes:
df_enroll_agg = (
    df_enroll
    .groupby(['Precinct', 'Year'], as_index=False)
    .agg({
        'Voter_Enrollment': 'sum',  # or 'mean' / 'max' / etc.
        'TotalEligible': 'sum'
    })
)

# Merge on (Precinct, Year)
df_final = pd.merge(
    df_merged,
    df_enroll_agg,
    on=['Precinct', 'Year'],
    how='left'
)

##############################################################################
# 6. SAVE / INSPECT THE FINAL MERGED DATA
##############################################################################
print(df_final.head(10))

# Optionally save to CSV for further analysis
df_final.to_csv("merged_for_clustering.csv", index=False)
```

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as ctx
from shapely.geometry import Point
from sklearn.cluster import KMeans
import ipywidgets as widgets
from IPython.display import display

# ---------------------------------------------------------------------
# 1. Load your data
# ---------------------------------------------------------------------
df = pd.read_csv('merged_for_clustering.csv')  # Replace with your CSV path

# Expected columns based on your data:
# ['VoterID', 'Year', 'TurnoutRate',
#  'Residence_Addresses_Latitude',
#  'Residence_Addresses_Longitude',
#  'Precinct', ...]

# ---------------------------------------------------------------------
# 2. Convert df to a GeoDataFrame in EPSG:4326 (lat/lon)
# ---------------------------------------------------------------------
geometry = [
    Point(lon, lat) for lon, lat in zip(
        df['Residence_Addresses_Longitude'],
        df['Residence_Addresses_Latitude']
    )
]
gdf = gpd.GeoDataFrame(df, geometry=geometry, crs="EPSG:4326")

# ---------------------------------------------------------------------
# 3. Define a function to cluster voters by precinct (and years)
#    and return a GeoDataFrame with cluster labels.
# ---------------------------------------------------------------------
def cluster_precinct_data(gdf, precinct, years=None, n_clusters=2, threshold_mode='overall'):
    """
    Cluster the voters within a given precinct (optionally for specific years)
    using their latitude/longitude, then compare each cluster’s turnout rate
    to a chosen threshold (either overall or precinct-level average).

    Returns a GeoDataFrame (subset) with a 'Cluster' column.
    """
    # Filter by precinct and optional years
    if years is not None and len(years) > 0:
        subset = gdf[(gdf['Precinct'] == precinct) & (gdf['Year'].isin(years))].copy()
    else:
        subset = gdf[gdf['Precinct'] == precinct].copy()

    if subset.empty:
        print(f"\nNo data found for precinct '{precinct}' with years={years}.")
        return None

    # Prepare numeric features for K-means (lat/lon)
    X = subset[['Residence_Addresses_Latitude', 'Residence_Addresses_Longitude']]

    # Run K-means
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    subset['Cluster'] = kmeans.fit_predict(X)

    # Calculate turnout stats
    overall_avg_turnout = subset['TurnoutRate'].mean()

    if threshold_mode == 'precinct':
        p_mask = (gdf['Precinct'] == precinct)
        precinct_avg_turnout = gdf.loc[p_mask, 'TurnoutRate'].mean()
        threshold = precinct_avg_turnout
    else:
        threshold = overall_avg_turnout

    cluster_turnout = subset.groupby('Cluster')['TurnoutRate'].mean()
    cluster_labels = cluster_turnout.apply(lambda x: 'High' if x >= threshold else 'Low')

    print(f"\nPrecinct: {precinct}")
    if years is not None:
        print(f"Years: {years}")
    print(f"Threshold mode: {threshold_mode} (Threshold = {threshold:.3f})")
    print("Average Turnout by Cluster:")
    for c_label in sorted(cluster_turnout.index):
        print(f"  Cluster {c_label}: {cluster_turnout[c_label]:.3f} ({cluster_labels[c_label]})")

    return subset

# ---------------------------------------------------------------------
# 4. Define a function to plot a clustered GeoDataFrame on a basemap
# ---------------------------------------------------------------------
def plot_clusters_on_map(gdf_clustered, precinct, years=None):
    """
    Takes a GeoDataFrame with a 'Cluster' column and plots the points on a basemap.
    """
    if gdf_clustered is None or gdf_clustered.empty:
        print(f"No data to plot for precinct '{precinct}' with years={years}.")
        return

    # Reproject to Web Mercator (EPSG:3857) for basemap compatibility
    gdf_webmerc = gdf_clustered.to_crs(epsg=3857)

    fig, ax = plt.subplots(figsize=(8, 6))
    gdf_webmerc.plot(
        ax=ax,
        column='Cluster',
        categorical=True,
        legend=True,
        alpha=0.7,
        markersize=10
    )

    # Add a basemap; if Stamen.TonerLite is not available, use OpenStreetMap.Mapnik
    ctx.add_basemap(
        ax,
        crs=gdf_webmerc.crs,
        source=ctx.providers.OpenStreetMap.Mapnik
    )

    title = f"Precinct: {precinct}"
    if years is not None:
        title += f" | Years: {years}"
    ax.set_title(title)
    ax.set_axis_off()  # Hide axis for a cleaner look
    plt.show()

# ---------------------------------------------------------------------
# 5. Define an interactive update function
# ---------------------------------------------------------------------
def update_map(precinct, years):
    """
    Update the clustering and map plot based on selected precinct and years.
    The 'years' parameter comes in as a tuple from the multi-select widget.
    """
    # Convert the tuple of selected years to a list of integers
    selected_years = list(years)
    print("Selected years:", selected_years)

    # Cluster data for the given precinct and years
    clustered_gdf = cluster_precinct_data(
        gdf,
        precinct=precinct,
        years=selected_years,
        n_clusters=2,
        threshold_mode='overall'  # or change to 'precinct' if desired
    )

    # Plot the clustered data on a basemap
    plot_clusters_on_map(clustered_gdf, precinct, years=selected_years)

# ---------------------------------------------------------------------
# 6. Create interactive widgets for precinct and years selection
# ---------------------------------------------------------------------
# Get the unique precincts from your data
unique_precincts = list(gdf['Precinct'].unique())
precinct_dropdown = widgets.Dropdown(options=unique_precincts, description="Precinct:")

# Create a multi-select widget for years (from 2015 to 2022)
year_select = widgets.SelectMultiple(
    options=sorted(range(2015, 2023)),
    value=(2015, 2022),
    description="Years:"
)

# Display the interactive controls with a button to update the map
widgets.interact_manual(update_map, precinct=precinct_dropdown, years=year_select)

```
