```python
import pandas as pd
import numpy as np
import textwrap
```

```python
df = pd.read_csv("126234_Deliverable.csv")
```

```python
pd.set_option('display.max_columns', 900)
df.head(5)
```

```python
import pandas as pd
from IPython.display import display, HTML

# Display the null percentage of each column from highest to lowest
null_percentage = df.isnull().mean().sort_values(ascending=False)
null_percentage = null_percentage[null_percentage > 0]
null_percentage = null_percentage * 100

# Convert the null percentage to a DataFrame for better formatting
null_percentage_df = null_percentage.reset_index()
null_percentage_df.columns = ['Column', 'Null Percentage']

# Convert the DataFrame to HTML and wrap it in a scrollable div
html = null_percentage_df.to_html(index=False)
scrollable_html = f"<div style='height: 400px; overflow-y: scroll;'>{html}</div>"

# Display the scrollable HTML
display(HTML(scrollable_html))

# Now 
```

```python
df.shape
```

```python
# Drop all columns with a null percentage of 100%
columns_to_drop = null_percentage[null_percentage == 100].index
df_cleaned = df.drop(columns=columns_to_drop)

# Display the cleaned DataFrame
df_cleaned.head(5)
```

```python
df_cleaned.shape
```

```python
# Display the null percentage of each column from highest to lowest for df_cleaned
null_percentage_cleaned = df_cleaned.isnull().mean().sort_values(ascending=False)
null_percentage_cleaned = null_percentage_cleaned[null_percentage_cleaned > 0]
null_percentage_cleaned = null_percentage_cleaned * 100

# Convert the null percentage to a DataFrame for better formatting
null_percentage_cleaned_df = null_percentage_cleaned.reset_index()
null_percentage_cleaned_df.columns = ['Column', 'Null Percentage']

# Convert the DataFrame to HTML and wrap it in a scrollable div
html_cleaned = null_percentage_cleaned_df.to_html(index=False)
scrollable_html_cleaned = f"<div style='height: 400px; overflow-y: scroll;'>{html_cleaned}</div>"

# Display the scrollable HTML
display(HTML(scrollable_html_cleaned))
```

```python
# remove rows that have empty values in Voters_Age and Voters_BirthDate

df_cleaned_age = df_cleaned.dropna(subset=['Voters_Age', 'Voters_BirthDate'])

df_cleaned_age.shape
```

```python

# Set display options to show all columns
pd.set_option('display.max_columns', None)

# Display all columns as a list
columns = df_cleaned_age.columns.tolist()

# Wrap the text to display columns in a readable format but also in alphabetical order
wrapped_columns = textwrap.fill(str(columns), width=100)
#print(wrapped_columns)
```

```python
# Display the null percentage of each column from highest to lowest for df_cleaned
null_percentage_cleaned = df_cleaned_age.isnull().mean().sort_values(ascending=False)
null_percentage_cleaned = null_percentage_cleaned[null_percentage_cleaned >= 0]
null_percentage_cleaned = null_percentage_cleaned * 100

# Convert the null percentage to a DataFrame for better formatting
null_percentage_cleaned_df = null_percentage_cleaned.reset_index()
null_percentage_cleaned_df.columns = ['Column', 'Null Percentage']

# Convert the DataFrame to HTML and wrap it in a scrollable div
html_cleaned = null_percentage_cleaned_df.to_html(index=False)
scrollable_html_cleaned = f"<div style='height: 400px; overflow-y: scroll;'>{html_cleaned}</div>"

# Display the scrollable HTML
#display(HTML(scrollable_html_cleaned))
```
